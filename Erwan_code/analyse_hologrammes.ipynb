## Reading and saving results

- **Best fits** lists selected parameters. **All z trials** lists individual fitting attempts.
- **Result figures**, under the figures and analysis log section, selects the image and particle to inspect.
- The reported chi² is the mean squared residual per pixel. **Improvement** compares
  this error with a uniform image at the crop's mean intensity; it is not parameter
  accuracy or a confidence percentage.
- Inspect experimental/fitted fringes, residual structure, parameter-bound warnings
  and ambiguity notices before accepting a measurement.

Candidate numbers are specific to each image; they do not track particles over time.
The default depth hints are 5, 15, 25 and 40 µm, within default depth bounds [0, 60] µm.
Automatic initialization may test additional values within the bounds.
HoloPy uses x for the vertical row axis and y for the horizontal column axis.

Results are saved in the indicated run folder. Use **Save profile** to retain settings
and selections. Update input and output paths when moving a profile to another computer.
Click **Stop** to cancel and wait until completed results have been saved.
