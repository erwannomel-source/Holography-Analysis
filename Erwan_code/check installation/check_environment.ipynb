# Check the installation

Open the supplied `Erwan_code` folder in VS Code or Jupyter and select the Python
kernel intended for `analyse_hologrammes.ipynb`. Run the cell below.

The report checks that exact Python environment and the supplied project files.
It tests scientific imports, a small HoloPy hologram, PNG generation and interface
creation. The default check installs no packages and runs no particle fit.

Keep this notebook and `check_environment_code.py` together in `check installation/`.
After the checks pass, open `../analyse_hologrammes.ipynb` using the same kernel.

````python
from pathlib import Path
import subprocess
import sys

working_folder = Path.cwd().resolve()
project_candidates = (
    working_folder / 'Erwan_code',
    working_folder,
    working_folder.parent,
)
project_folder = next(
    (folder for folder in project_candidates
     if (folder / 'hologram_functions.py').is_file()
     and (folder / 'hologram_analysis').is_dir()
     and (folder / 'check installation' / 'check_environment_code.py').is_file()),
    None,
)
if project_folder is None:
    raise FileNotFoundError(
        'Open the complete Erwan_code folder as your workspace, '
        'then open check installation/check_environment.ipynb.'
    )
checker = project_folder / 'check installation' / 'check_environment_code.py'
print(f'Checking this kernel: {sys.executable}', flush=True)
result = subprocess.run(
    [sys.executable, '-B', str(checker), '--python', sys.executable],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
)
print(result.stdout)
if result.returncode:
    raise RuntimeError('Some checks failed. Follow the report and the instructions below.')
````

## Install missing interface packages (optional)

```python
result = subprocess.run(

    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
)
print(result.stdout)
if result.returncode:
    raise RuntimeError('Package installation or recheck failed.')
```

This installs only missing interface packages in the current kernel's environment,
using recorded versions and constraints on packages already installed. It then
runs the checks again. **Restart the kernel after installation.**
